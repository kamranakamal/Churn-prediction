```python
from sklearn.model_selection import train_test_split
X = df.drop("Churn", axis=1)
y = df['Churn']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler


cat_col = ['Subscription Type', 'Contract Length', 'Gender']
num_col = [col for col in X_train.columns.tolist() if col not in cat_col]

cat_pipeline = Pipeline([
    ('ohe', OneHotEncoder(handle_unknown='ignore'))
])

num_pipline = Pipeline([
    ('scaler', StandardScaler())
])

preprocessor = ColumnTransformer([
    ('ohe', cat_pipeline, cat_col),
    ('scaler', num_pipline, num_col)
])

X_train_encoded = preprocessor.fit_transform(X_train)
X_test_encoded = preprocessor.transform(X_test)
```

```python
import joblib
joblib.dump(preprocessor, '../Models/preprocessor.pkl')
```

```python
from sklearn.metrics import accuracy_score, recall_score, f1_score, precision_score, confusion_matrix
def eval_fn(y_true, y_pred):
    acc = accuracy_score(y_true, y_pred)
    recall = recall_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred)
    confus_mat = confusion_matrix(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)
    print(f"Accuracy Score: {acc} | Recall: {recall} | Precision: { precision} | f1 score: {f1}")
    sns.heatmap(confus_mat,annot=True, fmt='d', cmap='Blues')
    return acc, recall, precision, f1
    
```

```python
from sklearn.linear_model import LogisticRegression
log_reg = LogisticRegression(class_weight={0:1, 1:1.2})
log_reg.fit(X_train_encoded,y_train)
y_pred_log = log_reg.predict(X_test_encoded)
eval_fn(y_test, y_pred_log)
```

```python
from sklearn.naive_bayes import MultinomialNB, GaussianNB
gaussin_nb = GaussianNB()
gaussin_nb.fit(X_train_encoded, y_train)
y_pred_gnb = gaussin_nb.predict(X_test_encoded)
eval_fn(y_test, y_pred_gnb)
```

```python
from xgboost import XGBClassifier
xgb = model = XGBClassifier(
    objective='binary:logistic',
    learning_rate=0.1,
    max_depth=6,
    n_estimators=100,
    subsample=0.8,
    colsample_bytree=0.8,
    reg_lambda=1,
    reg_alpha=0,
    eval_metric='logloss',
    random_state=42
)
xgb.fit(X_train_encoded,y_train)
y_pred_xgb = xgb.predict(X_test_encoded)
eval_fn(y_test, y_pred_xgb)
```

```python
import joblib
joblib.dump(xgb, "../Models/XGBClassifier.pkl")
```

```python
from sklearn.neighbors import KNeighborsClassifier
kn_clf = KNeighborsClassifier()
kn_clf.fit(X_train_encoded, y_train)
y_pred_kn = kn_clf.predict(X_test_encoded)
eval_fn(y_test, y_pred_kn)
```
